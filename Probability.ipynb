```python
import statistics
scores=[50,23,52,31,56,84]
mean  =statistics.mean(scores)
mode =statistics.mode(scores)
median =statistics.median(scores)
print(f'data={sorted(scores)}')
print(f'mean={mean:.2f}')
print(f'mode={mode}')
print(f'median={median}')
```

```python
data_range  = max(scores) - min(scores)
variance    = statistics.variance(scores)
std_dev     = statistics.stdev(scores)
print(f'Range      : {data_range}')
print(f'Variance   : {variance:.2f}')
print(f'Std Dev    : {std_dev:.2f}')
```

```python
data = [2, 3, 4, 1, 7, 8, 12, 11, 10]
n = len(data)
mean = sum(data) / n

deviation = [x - mean for x in data]
squared_dev = [d**2 for d in deviation]

variance = sum(squared_dev) / n
std_dev = variance**0.5


print(f'mean={mean:.2f}')
print("deviation=", [round(x, 2) for x in deviation])
print("squared deviation=", [round(x, 2) for x in squared_dev])
print(f"variance={variance:.2f}")
print(f"std_dev={std_dev:.2f}")
```

```python
total=8
p_getting_5=1/total
p_getting_even=4/total
p__getting_odd=1-p_getting_even
p_less_than_4=3/total
print(f'probability of getting 5 ={p_getting_5}')
print(f'probability of getting even ={p_getting_even}')
print(f'probability of getting odd ={p__getting_odd}')
print(f'probability of getting less than 4 ={p_less_than_4}')
```

```python
import random
random.seed(25)
n_rolls=5000
count_even=0
for _ in range(n_rolls):
  roll=random.randint(1,6)
  if roll%2==0:
    count_even+=1
experimental_probability = count_even / n_rolls
print(f'rolls={n_rolls}')
print(f'count of even={count_even}')
print(f'experimental probability of getting even ={experimental_probability:4f}')
print(f'thoritical probability = 0.5000')


```
